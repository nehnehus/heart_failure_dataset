```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
try:
    df1 = pd.read_csv('../Data/heart.csv')
    df2 = pd.read_csv('../Data/heart_disease.csv')
    print('Files loaded successfully')
except FileNotFoundError:
    print("Error: Make sure 'heart.csv' and 'heart_disease.csv' are in this exact folder.")
    
```

```python
df1.columns = df1.columns.str.lower().str.strip()
df2.columns = df2.columns.str.lower().str.strip()
```

```python
combined_data = pd.concat([df1, df2], ignore_index=True)
```

```python
combined_data['Patient_ID'] = range(1, len(combined_data) + 1)
```

```python
patients_table = pd.DataFrame()
patients_table['Patient_ID'] = combined_data['Patient_ID']

if 'age' in combined_data.columns:
    patients_table['Age'] = combined_data['age']
```
